# Operador OX

Parêmentros

```python
pai1 = [1, 2, 3, 4, 5, 6, 7, 8]
pai2 = [8, 7, 6, 5, 3, 4, 2, 1]
cut1, cut2 = 3, 6
```

Variáveis

```python
tamanho = len(pai1)
filho = [None] * tamanho
elementos_no_corte = set(pai1[cut1:cut2])
```

```python
filho
```

```python
elementos_no_corte
```

Copia dos genes do pai1

```python
filho[cut1:cut2] = pai1[cut1:cut2]
filho
```

Gera a sequência a partir do corte no pai2

```python
pai2_em_ordem_circular = [pai2[(cut2 + i) % tamanho] for i in range(tamanho)]
pai2_em_ordem_circular
```

Gera um vetor com os indices das posições que estao fora do corte

```python
posicoes_vazias = [
    (cut2 + i) % tamanho
    for i in range(tamanho)
    if (cut2 + i) % tamanho < cut1 or (cut2 + i) % tamanho >= cut2
]
posicoes_vazias
```

```python
idx_vaga = 0

for elemento in pai2_em_ordem_circular:
        # Se o elemento ainda não está no filho, insere na próxima vaga circular
        if elemento not in elementos_no_corte:
            pos_dest = posicoes_vazias[idx_vaga]
            filho[pos_dest] = elemento
            print(
                f"   Elemento {elemento} do Pai 2 -> Posição {pos_dest} | Estado: {filho}"
            )
            idx_vaga += 1
```

# Operador PMX

Parametros

```python
pai1 = [1, 5, 4, 7, 8, 2, 3, 6]
pai2 = [1, 8, 3, 4, 5, 6, 2, 7]
cut1, cut2 = 3, 6
```

Variáveis

```python
tamanho = len(pai1)
filho = [None] * tamanho
```

Copia a fatia do Pai 1 para o filho

```python
filho[cut1:cut2] = pai1[cut1:cut2]
filho
```

Cria a tabela de mapeamento do corte

```python
mapeamento = {}
for i in range(cut1, cut2):
    mapeamento[pai1[i]] = pai2[i]

mapeamento
```

 Preenche as posições fora do corte usando os elementos do Pai 2

```python
for i in range(tamanho):
    # Pula as posições centrais que já foram preenchidas
    if cut1 <= i < cut2:
        continue

    val = pai2[i]

    # Se o valor já está na fatia central do filho, resolve o conflito em cadeia
    while val in filho[cut1:cut2]:
        val_antigo = val
        val = mapeamento[val]
        print(f"   ⚠️ Conflito na pos {i}: {val_antigo} já existe -> Mapeado para {val}")

    filho[i] = val
    print(f"   Posição {i} preenchida com {val} -> Estado: {filho}")
```

# Operador CX

Parametros

```python
pai1 = [6,2,8,4,1,7,3,5]
pai2 = [3,7,5,1,6,4,2,8]
```

Variáveis

```python
tamanho = len(pai1)
filho = [None] * tamanho
```

Monta o índice invertido do pai1

```python
pos_no_pai1 = [None] * (tamanho + 1)
for i in range(tamanho):
    pos_no_pai1[pai1[i]] = i
    print(f"pos_no_pai1[{pai1[i]}] = {i}")

pos_no_pai1
```

Seleciona copias os genes do pai1, enquanto o pai2 seleciona os genes do pai1

```python
idx = 0
while filho[idx] == None:
    # Gene selecionado do pai1
    gene = pai1[idx]
    # Copia o gene para o filho
    filho[idx] = gene

    # Proximo gene do pai1 (é selecionado pelo pai2)
    novo_gene = pai2[idx]

    # Posicao do novo gene no pai1
    idx = pos_no_pai1[novo_gene]

    print(f"copia gene {gene} e seleciona o gene {novo_gene}")
```

```python
filho
```

Preenche os genes do pai2 no filho

```python
for i in range(tamanho):
    if filho[i] == None:
        filho[i] = pai2[i]

filho
```

# Operador GNX

Parêmetros

```python
pai1 = [1, 2, 3, 7, 8, 9, 4, 5, 6, 10, 11, 12]
pai2 = [4, 5, 6, 1, 2, 3, 10, 11, 12, 7, 8, 9]
cut_points = [3, 6, 9] 
```

Configuracao

```python
import random
random.seed(42)
```

Variáveis

```python
tamanho = len(pai1)
filho = [None] * tamanho

pontos = [0] + sorted(cut_points) + [tamanho]
fatias = [(pontos[i], pontos[i + 1]) for i in range(len(pontos) - 1)]
num_segmentos = len(fatias)
pais_fase1 = [random.choice([1, 2]) for _ in range(num_segmentos)]
pais_fase2 = [2 if p == 1 else 1 for p in pais_fase1]
```

Sorteia aleatoriamente qual pai fornece cada segmento

Para cada segmento ele copia o gene do pai sorteado se e somente o gene for valido

```python
def preencher_segmentos(filho, pai1, pai2, fatias, pais_por_segmento):
    tamanho = len(pai1)
    num_segmentos = len(fatias)
    
    for f in range(num_segmentos):
        inicio, fim = fatias[f]
        n_pai = pais_por_segmento[f]
        pai = pai1 if n_pai == 1 else pai2
    
        print(f"Pai{n_pai} sorteado")
        for i in range(inicio, fim):
            if filho[i] != None: continue
                
            gene = pai[i]
    
            gene_valido = True
            for j in range(tamanho):
                if gene == filho[j]:
                    gene_valido = False
                    print(f"\tGene {gene} do pai{n_pai} é invalido")
                    break
    
            if gene_valido:
                filho[i] = gene
                print(f"Copia o gene {gene} do pai{n_pai} para o filho")
        
```

Executa para a fase 1

```python
preencher_segmentos(filho, pai1, pai2, fatias, pais_fase1)
filho
```

Executa para a fase 2

```python
preencher_segmentos(filho, pai1, pai2, fatias, pais_fase2)
filho
```

Faz a varredura para saber qual gene ainda falta

```python
presente = [0] * (tamanho + 1)
for i in range(tamanho):
    if filho[i] is not None:
        presente[filho[i]] = 1

presente
```

Cria o vetor com os genes faltantes

```python
elementos_faltantes = []
for gene in range(1, tamanho + 1):
    if presente[gene] == 0:
        elementos_faltantes.append(gene)

elementos_faltantes
```

Embaralha o vetor para preencher os genes restantes de forma aleatória

```python
qtd_faltantes = len(elementos_faltantes)

if qtd_faltantes > 0:
    print("--- FASE 3: Reparo Aleatório ---")
    print(f"Elementos faltantes para alocar: {elementos_faltantes}")

    # Embaralhamento Fisher-Yates
    for i in range(qtd_faltantes - 1, 0, -1):
        r = random.randint(0, i)
        temp = elementos_faltantes[i]
        elementos_faltantes[i] = elementos_faltantes[r]
        elementos_faltantes[r] = temp

    print(f"Embaralhados: {elementos_faltantes}")

    idx_faltante = 0
    for i in range(tamanho):
        if filho[i] is None:
            val = elementos_faltantes[idx_faltante]
            filho[i] = val
            idx_faltante += 1
            print(f"\tPosição {i} -> Reparado com {val} | Estado: {filho}")
```
